```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Input data
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Expected output
y = np.array([0, 0, 0, 1])
```

```python
# Initialize weights and bias
weights = np.zeros(2)
bias = 0

learning_rate = 0.1
epochs = 10
```

```python
# Step activation function
def step(x):
    return 1 if x >= 0 else 0

# Training
for epoch in range(epochs):
    for i in range(len(X)):
        # Calculate output
        z = np.dot(X[i], weights) + bias
        prediction = step(z)

        # Calculate error
        error = y[i] - prediction

        # Update weights and bias
        weights = weights + learning_rate * error * X[i]
        bias = bias + learning_rate * error

print("Weights:", weights)
print("Bias:", bias)
```

```python
for i in range(len(X)):
    z = np.dot(X[i], weights) + bias
    prediction = step(z)

    print(X[i], "->", prediction)
```

```python
# Plot data points
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='bwr', s=100)

# Decision boundary
x1 = np.linspace(-0.5, 1.5, 100)
x2 = -(weights[0] * x1 + bias) / weights[1]

plt.plot(x1, x2, color='black', label='Decision Boundary')

plt.xlabel("Input 1")
plt.ylabel("Input 2")
plt.title("Perceptron - AND Gate")
plt.legend()
plt.grid()
plt.show()
```
